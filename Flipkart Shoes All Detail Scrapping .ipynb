```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time
import os
import requests
import pandas as pd

# Open Chrome
driver = webdriver.Chrome()


# Open Flipkart
driver.get("https://www.flipkart.com")

# Maximize window
driver.maximize_window()

# Wait for page to load
time.sleep(5)

# Find the search box
search_box = driver.find_element(By.NAME, "q")

# Type "shoes"
search_box.send_keys("shoes")

# Press Enter
search_box.send_keys(Keys.ENTER)

# Wait to see the results
time.sleep(5)
shoes = driver.find_elements(By.CSS_SELECTOR, ".bLCLBY.nr15la")

data = []

desktop = os.path.join(os.path.expanduser("~"), "Desktop")

image_folder = os.path.join(desktop, "Flipkart_Images")

os.makedirs(image_folder, exist_ok=True)

for i,shoe in enumerate(shoes[:40],start=1):

    name = shoe.find_element(By.CLASS_NAME,"Fo1I0b").text
    price = shoe.find_element(By.CLASS_NAME, "hZ3P6w").text
    image = shoe.find_element(By.CLASS_NAME,"MZeksS")
    image_url = image.get_attribute("src")
    safe_name = "".join(c for c in name if c not in r'\/:*?"<>|')
    filename = os.path.join(image_folder, f"{i}_{safe_name}.jpg")

    response = requests.get(image_url)

    if response.status_code == 200:
        image_data = response.content
        with open(filename, "wb") as f:
            f.write(image_data)
    else:
        print(f"Image {i} download failed.")
        continue

    

    data.append({
        "Name" :name,
        "Price":price,
        "image":filename
    })
            


df = pd.DataFrame(data)
excel_file = os.path.join(desktop, "Flipkart_Shoes.xlsx")
df.to_excel(excel_file, index=False)


print(f"Excel File Saved : {excel_file}")
print(f"Images Saved In : {image_folder}")

driver.quit()

print("Done!")


```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import requests
import os
import time

driver = webdriver.Chrome()

driver.get("https://books.toscrape.com/")

time.sleep(3)

books = driver.find_elements(By.CLASS_NAME, "product_pod")

data = []

os.makedirs("images", exist_ok=True)

for i, book in enumerate(books[:20], start=1):

    title = book.find_element(By.TAG_NAME, "h3").text
    price = book.find_element(By.CLASS_NAME, "price_color").text

    img = book.find_element(By.TAG_NAME, "img")
    img_url = img.get_attribute("src")

    img_data = requests.get(img_url).content

    filename = f"images/book_{i}.jpg"

    with open(filename, "wb") as f:
        f.write(img_data)

    data.append({
        "Title": title,
        "Price": price,
        "Image": filename
    })

df = pd.DataFrame(data)
df.to_excel("books.xlsx", index=False)

driver.quit()

print("Done!")
```
